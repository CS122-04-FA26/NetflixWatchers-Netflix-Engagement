## 0. Setup

### 0.1 Load Libraries

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import os
import kagglehub
%matplotlib inline
```

### 0.2. Load Dataset

```python
# Download latest version
path = kagglehub.dataset_download("zeyadmohamed26/netflix-customer-churn-and-engagement-analytics")

print("Path to dataset files:", path)
```

### 0.3 Insert Data into DataFrame

```python
df = pd.read_csv(path + "/netflix_customer_churn.csv")   # use the name from Step 1
df.head()
```

**Observations:** `customer_id` is a long random ID; will check if it is unique.

## 1. Inspect the Data

### 1.1 Identify Parameters

```python
# Dataset size
print("=== Dataset Size ===")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

# Column names and data types
print("\n=== Column Names & Data Types ===")
dtypes_df = df.dtypes.to_frame(name="dtype").reset_index()
dtypes_df.columns = ["column", "dtype"]
print(dtypes_df)

# Check if customer_id is unique
print("\n=== Unique ID Check ===")
print("Unique IDs:", df["customer_id"].nunique())
print("Total rows:", len(df))
print("customer_id is unique:", df["customer_id"].is_unique)
```

**Observations:** 
- 1 Unique ID (str): `customer_id` (It won't help explain churn, so it will be excluded from the model's features but kept to identify flagged subscribers)
- 6 categorical features (str): `gender`, `subscription_type`, `region`, `device`, `payment_method`, `favorite_genre`
- 7 numeric features (int64/float64): `age`, `watch_hours`, `last_login_days`, `monthly_fee`, `churned`, `number_of_profiles`, `avg_watch_time_per_day`

Overall, all data types look correct, so no type conversion is needed

### 1.2 Check for Duplicate/Missing Values

```python
# Check for fully duplicated rows
print("=== Duplicate Rows ===")
print("Duplicate rows:", df.duplicated().sum())

# Check for missing values
print("\n=== Missing Values ===")
print(df.isnull().sum())
print("\nTotal missing values:", df.isnull().sum().sum())
```

**Observations:** No duplicate rows or missing values in any column, so no rows or columns need to be dropped.

### 1.3 Descriptive Statistics

```python
# Counts for each qualitative column
print("=== Category Counts ===")
print()
for col in ["gender", "subscription_type", "region", "device", "payment_method", "favorite_genre"]:
    print("---", col, "---")
    print(df[col].value_counts().to_string())
    print()

# Summary statistics for quantitative columns
print("=== Summary Statistics ===")
display(df.describe())
```

**Observations:**
- Every categorical column is split almost evenly across its categories, which is expected since the dataset is synthetic. Real Netflix data would likely be more uneven. Noted as a limitation for the model.
- `churned` mean = 0.50, so it is balanced as about half of subscribers churned.
- `age` ranges 18-70 with the average being around 44.
- `watch_hours` is right-skewed since the mean (11.65) > the median (8.0). Also the max (110.4) > the 75th percentile (16.03) by a lot, so there are some heavy outliers.
- `last_login_days` ranges 0-60 with the mean ≈ median (30), so it's evenly spread.
- `monthly_fee` only has 3 values (8.99, 13.99, 17.99), so results match the 3 subscription plans (Basic, Standard, Premium)
- `number_of_profiles` ranges 1-5 with the average being 3.
- `avg_watch_time_per_day` ranges from 0.00-98.42 hrs, which is questionable since the Kaggle documentation says the range is 0.01-4.97 hrs. Some rows fall outside the documented range and will be checked in the cleaning section.
